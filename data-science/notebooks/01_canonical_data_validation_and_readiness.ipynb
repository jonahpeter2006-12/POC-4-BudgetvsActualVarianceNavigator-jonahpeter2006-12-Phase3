# Phase 3 ? Canonical Data Validation and Analytical Readiness

This notebook reviews the canonical dataset and the four generated Phase 3 assessment outputs.

**Important:** The dataset is synthetic and contains 18 original source observations represented by 72 canonical metric records.

## 1. Load the canonical dataset

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd()
DATA_PATH = ROOT / 'data' / 'canonical' / 'intelligence_data.csv'
OUTPUT_PATH = ROOT / 'data-science' / 'outputs'

df = pd.read_csv(DATA_PATH)
print('Rows:', len(df))
print('Columns:', len(df.columns))
display(df.head())
```

## 2. Review dataset structure

```python
display(df.dtypes.to_frame('Data type'))
display(df.isnull().sum().to_frame('Missing values'))
```

## 3. Review Phase 3 assessment outputs

```python
assessment_files = {
    'Canonical profile': 'canonical_profile.json',
    'Data quality': 'quality_assessment.json',
    'Representativeness': 'representativeness_assessment.json',
    'Analytical readiness': 'analytical_readiness.json'
}

for label, filename in assessment_files.items():
    path = OUTPUT_PATH / filename
    with path.open(encoding='utf-8') as file:
        result = json.load(file)
    print('\n' + label)
    print(json.dumps(result, indent=2)[:3000])
```

## 4. Analytical limitations

- All records are synthetic.
- There are only 18 original source observations.
- The 72 canonical records are not 72 independent transactions.
- Predictive reliability and real-world representativeness have not been established.
- Comparative and limited descriptive analysis are the supported analytical directions.
