## 1. Analytical definitions

The approved sign convention is:

- Variance = Actual ? Budget
- Variance percentage = (Actual ? Budget) / Budget ? 100
- Positive variance indicates overspend.
- Negative variance indicates underspend.
- Zero variance indicates on-budget performance.

Budget and Actual are the only measures used in variance calculations. Committed and Forecast remain separate contextual measures and are not added to either amount.

Group percentages are calculated from group-level Budget and Actual totals, not by averaging individual percentages.

```python
from pathlib import Path
import json
import subprocess
import sys

ROOT = Path.cwd()

if not (ROOT / "data" / "canonical" / "intelligence_data.csv").exists():
    raise FileNotFoundError(
        "Run this notebook from the project root directory."
    )

DATASET = ROOT / "data" / "canonical" / "intelligence_data.csv"
SCRIPTS = ROOT / "data-science" / "scripts"
OUTPUTS = ROOT / "data-science" / "outputs"

print("Project root:", ROOT)
print("Canonical dataset:", DATASET)
print("Dataset exists:", DATASET.exists())
print("Approved track: Track A ? Comparative Intelligence")
```

## 2. Execute the comparative intelligence method

The analytical runner reads the canonical dataset, pairs the four financial measures for each independent source observation, and generates overall and monthly comparative findings.

```python
runner = SCRIPTS / "run_analytical_track.py"

run_result = subprocess.run(
    [sys.executable, str(runner)],
    cwd=ROOT,
    capture_output=True,
    text=True
)

print(run_result.stdout)
if run_result.stderr:
    print(run_result.stderr)

if run_result.returncode != 0:
    raise RuntimeError("Analytical track execution failed.")

results_path = OUTPUTS / "intelligence_results.json"
results_payload = json.loads(results_path.read_text(encoding="utf-8"))

print("Generated findings:", len(results_payload["results"]))
print("Method version:", results_payload["method_version"])
print("Dataset version:", results_payload["dataset_version"])
```

## 3. Independently validate analytical findings

The independent validation script checks the generated findings against the canonical dataset, including calculations, evidence traceability, measure separation, schema, classification, limitations, and ranking eligibility.

```python
validator = SCRIPTS / "validate_analytical_track.py"

validation_result = subprocess.run(
    [sys.executable, str(validator)],
    cwd=ROOT,
    capture_output=True,
    text=True
)

print(validation_result.stdout)
if validation_result.stderr:
    print(validation_result.stderr)

if validation_result.returncode != 0:
    raise RuntimeError("Independent analytical validation failed.")
```

## 4. Export structured intelligence outputs

The export step produces the summary, validation metrics, and weak-case review. These outputs retain the approved method and dataset versions.

```python
exporter = SCRIPTS / "export_intelligence_results.py"

export_result = subprocess.run(
    [sys.executable, str(exporter)],
    cwd=ROOT,
    capture_output=True,
    text=True
)

print(export_result.stdout)
if export_result.stderr:
    print(export_result.stderr)

if export_result.returncode != 0:
    raise RuntimeError("Intelligence output export failed.")
```

## 5. Review generated findings and validation evidence

The following checks confirm the generated output counts and validation status. Monthly findings with fewer than three independent source observations remain descriptive and unranked.

```python
summary = json.loads(
    (OUTPUTS / "intelligence_summary.json").read_text(encoding="utf-8")
)
metrics = json.loads(
    (OUTPUTS / "validation_metrics.json").read_text(encoding="utf-8")
)
weak_review = json.loads(
    (OUTPUTS / "weak_case_review.json").read_text(encoding="utf-8")
)

print("Total findings:", summary["total_findings"])
print("Overall findings:", summary["overall_finding_count"])
print("Monthly findings:", summary["monthly_finding_count"])
print("Overall ranked findings:", summary["overall_ranked_finding_count"])
print("Monthly ranked findings:", summary["monthly_ranked_finding_count"])
print("Validation status:", metrics["overall_status"])
print("Validation checks:", (metrics["passed_count"] + metrics["failed_count"]))
print("Failed checks:", metrics["failed_count"])
print("Weak cases:", weak_review["weak_case_count"])

assert summary["total_findings"] == 64
assert summary["overall_finding_count"] == 16
assert summary["monthly_finding_count"] == 48
assert summary["monthly_ranked_finding_count"] == 0
assert metrics["overall_status"] == "PASS"
assert metrics["failed_count"] == 0
assert weak_review["weak_case_count"] == 48

print("Notebook review assertions: PASS")
```

## 6. Limitations and interpretation

- The canonical dataset contains 18 independent synthetic source observations.
- The 72 canonical rows represent four financial measures per source observation, not 72 independent transactions.
- Monthly findings with fewer than three independent observations are descriptive and marked insufficient evidence.
- The minimum evidence threshold is a screening rule, not a statistical representativeness guarantee.
- Forecast values are illustrative and are not validated predictive outputs.
- The results do not establish population-level conclusions.
- No predictive model, anomaly detector, causal inference, or autonomous financial decision system is developed.
